## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
print()

print("Average impressions (90d):", round(df["impressions_90d"].mean(), 2))
print("Average sessions (90d):", round(df["sessions_90d"].mean(), 2))
print()

print(df["trend_direction"].value_counts())
```

Out of 30,000 pages, 16,262 (54%) show a declining trend, and average impressions (5,200) far exceed average sessions (37) — confirming a large visibility-to-traffic gap that manual review can't address at scale, making prioritization essential.
